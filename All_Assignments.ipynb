Are lation R on a set A is given as a set of ordered pairs.Write a program to check whether R is:
•Reflexive
• Symmetric
• Transitive

```python
# Input set A
A = set(map(int, input("Enter elements of set A (space-separated): ").split()))

# Input relation R as ordered pairs
n = int(input("Enter number of ordered pairs in relation R: "))

R = set()
print("Enter pairs (a b):")
for _ in range(n):
    a, b = map(int, input().split())
    R.add((a, b))

# -------------------------------
# Check Reflexive
reflexive = True
for a in A:
    if (a, a) not in R:
        reflexive = False
        break

# -------------------------------
# Check Symmetric
symmetric = True
for (a, b) in R:
    if (b, a) not in R:
        symmetric = False
        break

# -------------------------------
# Check Transitive
transitive = True
for (a, b) in R:
    for (c, d) in R:
        if b == c and (a, d) not in R:
            transitive = False
            break
    if not transitive:
        break

# -------------------------------
# Output results
print("\nRelation Properties:")
print("Reflexive:", "Yes" if reflexive else "No")
print("Symmetric:", "Yes" if symmetric else "No")
print("Transitive:", "Yes" if transitive else "No")
```

22.Write a program to display a relation using its adjacency matrix.

```python
# Input set A
A = list(map(int, input("Enter elements of set A (space-separated): ").split()))
n = len(A)

# Create index mapping (element → index)
index = {A[i]: i for i in range(n)}

# Initialize adjacency matrix with 0
M = [[0]*n for _ in range(n)]

# Input relation R
m = int(input("Enter number of ordered pairs in relation R: "))
print("Enter pairs (a b):")

for _ in range(m):
    a, b = map(int, input().split())
    if a in index and b in index:
        i = index[a]
        j = index[b]
        M[i][j] = 1

# Display adjacency matrix
print("\nAdjacency Matrix:")
for row in M:
    print(*row)
```

Write a Python program to simplify the following Boolean expression using the boolalg module
of sympy library:
F(A,B,C) = (A∧B)∨(¬A∧C)∨(B∧C)

```python
from sympy import symbols
from sympy.logic.boolalg import And, Or, Not, simplify_logic

# Define variables
A, B, C = symbols('A B C')

# Define Boolean expression
F = Or(And(A, B), And(Not(A), C), And(B, C))

# Simplify expression
F_simplified = simplify_logic(F, form='dnf')

# Output
print("Original Expression:", F)
print("Simplified Expression:", F_simplified)
```

Write a Python program to check whether the following logical statement is a tautology: (P ∨Q)∨¬P

```python
from sympy import symbols
from sympy.logic.boolalg import Or, Not, simplify_logic

# Define variables
P, Q = symbols('P Q')

# Define the logical expression
expr = Or(Or(P, Q), Not(P))   # (P ∨ Q) ∨ ¬P

# Simplify expression
simplified_expr = simplify_logic(expr)

# Check if tautology
if simplified_expr == True:
    print("The statement is a TAUTOLOGY")
else:
    print("The statement is NOT a tautology")

# Print simplified form
print("Simplified Expression:", simplified_expr)
```

Ask user to specify the value for m and n where m,n ∈ N. Write a Python program that takes a
number in the base-m system and converts it to the base-n system.

```python
# Function to convert any base to decimal
def to_decimal(num_str, base):
    return int(num_str, base)

# Function to convert decimal to any base
def from_decimal(num, base):
    digits = "0123456789ABCDEFGHIJKLMNOPQRSTUVWXYZ"

    if num == 0:
        return "0"

    result = ""
    while num > 0:
        result = digits[num % base] + result
        num //= base

    return result

# -------------------------------
# User Input
m = int(input("Enter base m (original base): "))
n = int(input("Enter base n (target base): "))
num = input(f"Enter number in base-{m}: ").upper()

# Convert base-m → decimal
decimal = to_decimal(num, m)

# Convert decimal → base-n
result = from_decimal(decimal, n)

# Output
print(f"\nNumber in base-{n}:", result)
```

Write a Python program to:
(a) Compute P(X = 25) for 50 fair coin tosses.
(b) Print probabilities for k = 20 to 30.
(c) Verify that sum of all probabilities equals 1

```python
import math

n = 50
p = 0.5

# Function to compute binomial probability
def binomial_prob(n, k, p):
    return math.comb(n, k) * (p**k) * ((1-p)**(n-k))

# -------------------------------
# (a) Compute P(X = 25)
px_25 = binomial_prob(n, 25, p)
print("(a) P(X = 25) =", px_25)

# -------------------------------
# (b) Print probabilities for k = 20 to 30
print("\n(b) Probabilities for k = 20 to 30:")
for k in range(20, 31):
    print(f"P(X = {k}) = {binomial_prob(n, k, p)}")

# -------------------------------
# (c) Verify sum of all probabilities = 1
total_prob = sum(binomial_prob(n, k, p) for k in range(n+1))
print("\n(c) Sum of all probabilities =", total_prob)
```

Write a Python program to:
(a) Compute P(X = k) for k = 0to 8.
(b) Print probability table.
(c) Plot the PMF.

```python
import math
import matplotlib.pyplot as plt

n = 8
p = 0.2

# Function for binomial probability
def binomial_prob(n, k, p):
    return math.comb(n, k) * (p**k) * ((1-p)**(n-k))

# -------------------------------
# (a) Compute P(X = k) for k = 0 to 8
probabilities = []
print("(a) P(X = k):")
for k in range(n + 1):
    prob = binomial_prob(n, k, p)
    probabilities.append(prob)
    print(f"P(X = {k}) = {prob:.5f}")

# -------------------------------
# (b) Print probability table
print("\n(b) Probability Table:")
print("k\tP(X=k)")
for k, prob in enumerate(probabilities):
    print(f"{k}\t{prob:.5f}")

# -------------------------------
# (c) Plot PMF
k_values = list(range(n + 1))

plt.figure()
plt.bar(k_values, probabilities)
plt.xlabel("k (Number of errors)")
plt.ylabel("P(X = k)")
plt.title("Binomial PMF (n=8, p=0.2)")
plt.show()
```

 Adie is rolled 200 times with frequencies:
(20, 35,40,45,30,30)
Write a Python program to:
(a) Construct empirical PMF.
(b) Compute empirical mean.
(c) Compute empirical variance.
(d) Print results clearly.

```python
# Outcomes and their observed frequencies
outcomes = [1, 2, 3, 4, 5, 6]
freq = [20, 35, 40, 45, 30, 30]

total = sum(freq)

# -------------------------------
# (a) Empirical PMF
pmf = [f / total for f in freq]

print("(a) Empirical PMF:")
print("x\tP(X=x)")
for x, p in zip(outcomes, pmf):
    print(f"{x}\t{p:.4f}")

# -------------------------------
# (b) Empirical Mean
mean = sum(x * p for x, p in zip(outcomes, pmf))
print("\n(b) Empirical Mean =", round(mean, 4))

# -------------------------------
# (c) Empirical Variance
variance = sum(((x - mean) ** 2) * p for x, p in zip(outcomes, pmf))
print("(c) Empirical Variance =", round(variance, 4))

# -------------------------------
# (d) Summary
print("\n(d) Summary:")
print(f"Total trials = {total}")
print(f"Mean ≈ {mean:.4f}")
print(f"Variance ≈ {variance:.4f}")
```

5. Toss 10 coins.
Write a Python program to:
(a) Compute theoretical probability of exactly 5 heads.
(b) Simulate 100,000 trials.
(c) Compare theoretical and simulated result.

```python
import math
import random

n = 10
p = 0.5
trials = 100000

# -------------------------------
# (a) Theoretical probability
theoretical = math.comb(n, 5) * (p**5) * ((1-p)**5)
print("(a) Theoretical P(X=5) =", theoretical)

# -------------------------------
# (b) Simulation
count_5_heads = 0

for _ in range(trials):
    heads = sum(random.randint(0,1) for _ in range(n))
    if heads == 5:
        count_5_heads += 1

simulated = count_5_heads / trials
print("(b) Simulated P(X=5) =", simulated)

# -------------------------------
# (c) Comparison
print("\n(c) Comparison:")
print(f"Theoretical = {theoretical:.5f}")
print(f"Simulated   = {simulated:.5f}")
```

8.Write a Python program to compute the determinant of the matrix
A= [[2 1],[6 3]]
and determine whether the system Ax=b has a unique solution.

```python
import numpy as np

# Define matrix A
A = np.array([[2, 1],
              [6, 3]])

# Compute determinant
det_A = np.linalg.det(A)

print("Determinant of A =", det_A)

# Check uniqueness
if abs(det_A) < 1e-10:
    print("System does NOT have a unique solution (matrix is singular).")
else:
    print("System has a unique solution.")
```

 Generate a random 5×5 matrix in Python using NumPy,solve the system Ax=b,and verify the
solution by computing

```python
import numpy as np

# -------------------------------
# Generate random 5x5 matrix A
A = np.random.randint(1, 10, (5, 5))

# Generate random vector b
b = np.random.randint(1, 10, 5)

# -------------------------------
# Solve Ax = b
x = np.linalg.solve(A, b)

# -------------------------------
# Verify solution (compute Ax)
b_check = np.dot(A, x)

# -------------------------------
# Print results
print("Matrix A:\n", A)
print("\nVector b:\n", b)

print("\nSolution x:\n", x)

print("\nComputed Ax:\n", b_check)

print("\nOriginal b:\n", b)

# Check if Ax ≈ b
if np.allclose(b, b_check):
    print("\nVerification: Ax ≈ b (Correct solution)")
else:
    print("\nVerification failed")
```
